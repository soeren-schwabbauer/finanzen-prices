```python
from databricks.sdk import WorkspaceClient
from getpass import getpass
```

```python

w = WorkspaceClient()

w.secrets.create_scope(scope="finance")

api_key = getpass("Alpha Vantage API-Key: ")

w.secrets.put_secret(
    scope="finance",
    key="alpha-vantage",
    string_value=api_key
)

del api_key
```
